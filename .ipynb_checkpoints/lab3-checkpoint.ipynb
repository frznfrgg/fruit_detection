# Реализация и тестирование SSIM 
[Correlation Maximized Structural Similarity Loss for Semantic Segmentation](https://arxiv.org/abs/1910.08711)

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class SSIMLoss(nn.Module):
    def __init__(self, window_size=11, size_average=True):
        super(SSIMLoss, self).__init__()
        self.window_size = window_size
        self.size_average = size_average
        self.channel = 1
        self.window = self.create_window(window_size, self.channel)

    def gaussian(self, window_size, sigma):
        import math
        gauss = torch.Tensor([math.exp(-(x - window_size//2)**2/float(2*sigma**2)) for x in range(window_size)])
        return gauss / gauss.sum()

    def create_window(self, window_size, channel):
        _1D_window = self.gaussian(window_size, 1.5).unsqueeze(1)
        _2D_window = _1D_window.mm(_1D_window.t()).float().unsqueeze(0).unsqueeze(0)
        window = _2D_window.expand(channel, 1, window_size, window_size).contiguous()
        return window

    def forward(self, output, target):
        channel = output.shape[1]

        if channel == self.channel and self.window.data.type() == output.data.type():
            window = self.window
        else:
            window = self.create_window(self.window_size, channel)
            window = window.to(output.device).type(output.dtype)
            self.window = window
            self.channel = channel

        # сглаживание output и target
        mu1 = F.conv2d(output, window, padding=self.window_size//2, groups=channel)
        mu2 = F.conv2d(target, window, padding=self.window_size//2, groups=channel)

        mu1_sq = mu1.pow(2)
        mu2_sq = mu2.pow(2)
        mu1_mu2 = mu1 * mu2

        # отклонение (его аналог) по формуле: матожидание квадрата величины - квадрат матожидания
        # см фрагмент статьи №1
        sigma1_sq = F.conv2d(output * output, window, padding=self.window_size//2, groups=channel) - mu1_sq
        sigma2_sq = F.conv2d(target * target, window, padding=self.window_size//2, groups=channel) - mu2_sq
        sigma12 = F.conv2d(output * target, window, padding=self.window_size//2, groups=channel) - mu1_mu2

        C1 = 0.01**2
        C2 = 0.03**2

        # см фргамент статьи №2
        ssim_map = ((2 * mu1_mu2 + C1) * (2 * sigma12 + C2)) / ((mu1_sq + mu2_sq + C1) * (sigma1_sq + sigma2_sq + C2))

        if self.size_average:
            return 1.0 - ssim_map.mean()
        else:
            return 1.0 - ssim_map.mean(1).mean(1).mean(1)
```

# Испытаем SSIM + focal loss

```python
# кастомные датасет и u-net
from MLtools import ImageData, SC_Model

from torch.utils.data import DataLoader, random_split
from tqdm.notebook import tqdm
import numpy as np
from torchvision.ops import sigmoid_focal_loss
```

```python
image_folder = "ext_data"
xml_path = "annotations.xml"

dataset = ImageData(image_folder, xml_path, label_mod = False)
train_data, val_data = random_split(dataset, [9665, 100])

len(dataset)
```

```python
BS = 5

train_loader = DataLoader(train_data, batch_size=BS, shuffle=True)
val_loader = DataLoader(val_data, batch_size=BS, shuffle=False)
```

```python
def loss_SC_Model(output, target):
    # ВНИМАНИЕ! ssim_loss - глобальная переменная
    return sigmoid_focal_loss(output, target, reduction = 'mean') + ssim_loss(output, target)
```

```python
device = 'cuda:0' if torch.cuda.is_available() else 'cpu'
print(device)

model = SC_Model().to(device)
criterion = loss_SC_Model

train_losses = []
val_losses   = []
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr = 0.0002)
```

```python
n_epochs = 4

for epoch in tqdm(range(n_epochs)):
    model.train()
    train_losses_per_epoch = []
    val_losses_per_epoch   = []
    
    for batch, targets in train_loader:
        optimizer.zero_grad()
        
        outputs = model(batch.float().to(device))
        loss = criterion(outputs, targets.unsqueeze(1).float().to(device))
        loss.backward()
        optimizer.step()
        
        train_losses_per_epoch.append(loss.item())
    
    train_losses.append(np.mean(train_losses_per_epoch))
    train_losses_per_epoch = []
    
    model.eval()
    for batch, targets in val_loader:
        with torch.no_grad():
            outputs = model(batch.float().to(device))
            loss = criterion(outputs, targets.unsqueeze(1).float().to(device))
            val_losses_per_epoch.append(loss.item())
    
    val_losses.append(np.mean(val_losses_per_epoch))
    val_losses_per_epoch = []
```

```python
plt.plot(train_losses, label='train')
plt.plot(val_losses, label='validation')
plt.grid(True, linestyle='--')
plt.legend()
plt.title("Losses")
```

```python
N = 2
with torch.no_grad():
    for batch, target in val_loader:
        output = model(batch.float().to(device))
        
        ImageData.draw_appled_mask(batch[N], output[N,0].to('cpu'), mask_koef=0.5)
        break
```

# Сравним SSIM + focal и BCE + DICE

```python
def mean_IoU(model, val_loader, device=device, threshold=0.5):
    model.eval()
    total_iou = 0.0
    num_samples = 0
    SMOOTH = 1e-6 

    with torch.no_grad():
        for images, targets in val_loader:
            images = images.float().to(device)
            targets = targets.float().to(device)

            outputs = model(images)  # [B, 1, H, W]
            predicted_masks = (outputs.squeeze(1) > threshold).float()

            for pred, target in zip(predicted_masks, targets):
                pred_flat = pred.view(-1)
                target_flat = target.view(-1)

                intersection = (pred_flat * target_flat).sum()
                union = pred_flat.sum() + target_flat.sum() - intersection

                iou = (intersection + SMOOTH) / (union + SMOOTH)
                total_iou += iou.item()
                num_samples += 1

    mean_iou = total_iou / num_samples if num_samples > 0 else 0
    return mean_iou
```

### BCE + DICE

Обучение в тех же условиях, но с потерями вида
~~~
def DICE(pred, target):
    smooth = 1.0
    bs = pred.size(0)
    m1 = pred.view(bs, -1) 
    m2 = target.view(bs, -1)
    intersection = (m1 * m2).sum()

    return (2.0 * intersection + smooth) / (m1.sum() + m2.sum() + smooth)

def loss_SC_Model(output, target):
    BCE = torch.nn.BCELoss()
    return BCE(output, target) - 0.4*torch.log(DICE(output, target))
~~~
