```python
from PIL import Image
import os
```

```python
# усечение фотографий до квадратных
path0 = "raw_data"
to    = "croped_data"

dirs = os.listdir(path0)
s = 251

for dir in dirs:
    if dir != ".DS_Store":
        pics = os.listdir(f'{path0}/{dir}')
        for pic in pics:
            if pic != ".DS_Store":
                with Image.open(f'{path0}/{dir}/{pic}') as im:
                    width, height = im.size
                    sq = None
                    # вертикальная фотография - обрезаем низ до квадратного состояния
                    if height > width:
                        sq = im.crop((0,0,width,width))
                    # горизонтальная фотография - обрезаем края
                    elif height < width:
                        if (width - height)%2 == 0:
                            x = (width - height)//2
                            sq = im.crop((x,0,width-x,height))
                        else:
                            x  = (1 + width - height)//2
                            sq = im.crop((x,0,width-x,height))
                    # квадратная фотография - не делаем ничего
                    else:
                        sq = im
                        
                    # сжимем до 64х64 и сохраняем
                    try:
                        sq = sq.resize((640,640),Image.LANCZOS)
                        sq.save(f'{to}/{s}.jpg',quality=95)
                        s += 1
                    except:
                        print(f'{path0}/{dir}/{pic} is bugged')
            else:
                pass
    else:
        pass
```
